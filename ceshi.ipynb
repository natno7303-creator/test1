```python

def hello(name):
    print("hello ",name)
hello("zcc")

def introduce(name,age):
    print("我是",name,"今年",age,"岁")
introduce("zcc",22)

def add(a,b):
    return a+b
print(add(1,11))

def guess(x,answer=5):
    if x > answer:
        return "太大了"
    if x < answer:
        return "太小了"
    else:
        return "猜对了"
nums=[4,5,7]
for num in nums:
    print(num)
    print(guess(num))
```
