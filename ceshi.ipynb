```python
class Hello:
    count = 0
    def __init__(self,name):
        self.name=name
        Hello.count += 1  #每增加一个对象，count+1

    def hi(self):
        print("你好",self.name)

    def bye(self):
        print("再见",self.name)
    
    @classmethod
    def total(cls):
        print("当前共有",cls.count,"个对象")
class WarmHello(Hello):
    def remind(self):
        print(self.name,"说：路上小心")

hello_1 = Hello("11")
hello_1.hi()
hello_66=WarmHello("66")
hello_66.bye()
hello_66.remind()
Hello.total()
```

```python
class CookWare():  #父类
    def cook(self):
        print("正在做饭")

class Pot(CookWare):  #子类
    def cook(self):
        super().cook()  #先执行父类的方法
        print("正在用锅做饭")
    def make_soup(self):
        print("正在用锅煮汤")

class FryingPan(CookWare):  #子类
    def fry_egg(self):
        print("正在用平底锅煎蛋")

pot=Pot()
pan=FryingPan()
pot.cook()
pot.make_soup()
pan.cook()
pan.fry_egg()

print(type(pot) == Pot)  #pot的类型属于Pot
print(isinstance(pot,CookWare))  #pot也属于CookWare

def staring_cook(tool):
    tool.cook()
staring_cook(pot)
```

```python
while True: 
    try:
        cups = int(input("想要几杯咖啡？"))
        if cups <= 0:
            print("至少点1杯哦")
        else:
            print("这是你的",cups,"杯咖啡")
            break
    except ValueError:
        print("需要输入整数哦")

'''
1. ValueError - 值错误（类型正确但值不合适）
int("abc")          # ValueError
int("")             # ValueError

2. TypeError - 类型错误（操作类型不匹配）
"hello" + 123       # TypeError

3. ZeroDivisionError - 除以零错误
10 / 0              # ZeroDivisionError

4. FileNotFoundError - 文件不存在
open("不存在的文件.txt")  # FileNotFoundError

5. IndexError - 索引越界
[1, 2, 3][10]       # IndexError

6. KeyError - 字典键不存在
{"name": "兔娘"}["age"]  # KeyError

7. AttributeError - 属性不存在
"hello".nonexistent()   # AttributeError

8. NameError - 变量未定义
print(undefined_var)    # NameError

9. ImportError - 导入失败
import non_existent_module  # ModuleNotFoundError（ImportError 的子类）

10. KeyboardInterrupt - 用户按 Ctrl+C
（用户中断程序）
'''
```

```python
customers = [
    {"index":"1","status":"等待"},
    {"index":"2","status":"等待"},
    {"index":"3","status":"等待"},
]

def sever_customer(customer):
    customer["status"] = "取餐"
    print(f"第{customer['index']}名顾客可以取餐")

for customer in customers:
    print(f"第{customer['index']}名顾客正在等待")
    sever_customer(customer)
```
